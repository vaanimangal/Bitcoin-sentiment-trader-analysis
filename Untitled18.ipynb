# Bitcoin Market Sentiment vs Trader Performance Analysis

## Introduction

This project analyzes the relationship between Bitcoin market sentiment and trader performance using historical trading data from Hyperliquid along with the Bitcoin Fear & Greed Index.

The objective of this analysis is to understand how market emotions such as Fear, Greed, Extreme Fear, and Extreme Greed influence trading activity, profitability, win rates, and risk-taking behavior.

The project combines two datasets:

1. Bitcoin Fear & Greed sentiment dataset
2. Historical trader transaction dataset from Hyperliquid

The analysis includes:

* Data cleaning and preprocessing
* Dataset merging based on dates
* Profitability analysis
* Win rate analysis
* Trading activity analysis
* Trade size and risk behavior analysis
* Correlation analysis
* Visualization of key insights

The findings from this project help uncover behavioral trading patterns and provide insights into how market sentiment impacts trader decisions and overall market dynamics.


## Importing Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot')
```

## Loading Datasets

### Dataset Description

This project uses two datasets:

1. Bitcoin Fear & Greed Index Dataset

* Contains daily market sentiment classifications such as Fear, Greed, Extreme Fear, and Extreme Greed.

2. Hyperliquid Historical Trader Dataset

* Contains trader-level transaction data including execution price, trade size, leverage, coin traded, and closed profit/loss values.

These datasets were merged using date information to analyze the relationship between market sentiment and trader performance.


##Loading Files

```python
trades = pd.read_csv('historical_data.csv')
sentiment = pd.read_csv('fear_greed_index.csv')
```

##Check Column Names

```python
print(trades.columns)
print(sentiment.columns)
```

##Preview Data

```python
sentiment.head()
```

##Check Missing Values

```python
trades.isnull().sum()
```

Check Missing Values

```python
sentiment.isnull().sum()
```

##Convert Dates

```python
trades['Timestamp IST'] = pd.to_datetime(
    trades['Timestamp IST'],
    format='%d-%m-%Y %H:%M'
)

trades['date'] = trades['Timestamp IST'].dt.date

sentiment['date'] = pd.to_datetime(
    sentiment['date']
).dt.date
```

```python
trades[['Timestamp IST', 'date']].head()
```

##Merging the Datasets

```python
merged = pd.merge(
    trades,
    sentiment[['date', 'classification']],
    on='date',
    how='left'
)
```

##Check Merge

```python
merged[['date', 'Closed PnL', 'classification']].head()
```

##First Insight

```python
merged.groupby('classification')['Closed PnL'].mean()
```

##Creating First Visualization

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x='classification',
    y='Closed PnL',
    data=merged
)

plt.title('Trader Profitability vs Market Sentiment')
plt.xticks(rotation=45)

plt.show()
```

 Observation

The boxplot shows that trader profitability varies significantly across different market sentiment conditions. Extreme Greed periods produced the highest positive profit outliers, indicating strong bullish momentum and profitable trading opportunities. Fear and Greed markets also showed higher volatility in profit and loss outcomes compared to Neutral conditions.


##Trading Activity

```python
merged['classification'].value_counts()
```

##Win Rate Analysis

```python
merged['profit_trade'] = merged['Closed PnL'] > 0
```

##Win Rate Analysis

```python
merged.groupby('classification')['profit_trade'].mean() * 100
```

Observation

The win-rate analysis indicates that traders had the highest percentage of profitable trades during Extreme Greed periods. In contrast, Extreme Fear conditions showed lower win rates, highlighting the increased uncertainty and risk associated with panic-driven markets.


##Trading Activity

```python
activity = merged['classification'].value_counts()

print(activity)
```

##Visualization of Trading Activity

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x='classification',
    data=merged,
    order=merged['classification'].value_counts().index
)

plt.title('Trading Activity Across Market Sentiments')
plt.xticks(rotation=45)

plt.show()
```

Observation

The trading activity analysis shows that traders were most active during Fear market conditions. This suggests that volatile and uncertain markets encourage increased participation and short-term trading opportunities. Extreme Fear periods recorded comparatively lower trading activity.


##Trade Size and Risk Appetite Analysis

```python
merged.groupby('classification')['Size USD'].mean()
```

##Visualization for Trade Size

```python
plt.figure(figsize=(10,5))

sns.barplot(
    x='classification',
    y='Size USD',
    data=merged
)

plt.title('Average Trade Size by Market Sentiment')
plt.xticks(rotation=45)

plt.show()
```

Observation

The analysis reveals that traders placed the largest average trade sizes during Fear market conditions. This indicates increased risk-taking behavior and aggressive positioning during periods of uncertainty and volatility in the crypto market.


##Correlation Analysis

```python
numeric_cols = ['Execution Price', 'Size USD', 'Closed PnL']

correlation = merged[numeric_cols].corr()

print(correlation)
```

##Correlation Heatmap

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Between Trading Metrics')

plt.show()
```

 Observation

The correlation heatmap helps identify relationships between important trading variables such as execution price, trade size, and closed profit/loss. The analysis suggests that some trading metrics have weak to moderate relationships, highlighting the complex nature of trading performance in crypto markets.


##Top Performing Coins

```python
top_coins = merged.groupby('Coin')['Closed PnL'].mean().sort_values(ascending=False).head(10)

print(top_coins)
```

##Asset Performance Analysis

```python
top_coins.plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Top Coins by Average Profitability')
plt.ylabel('Average Closed PnL')

plt.show()
```

Observation

The asset performance analysis shows that profitability varied across different cryptocurrencies. Certain coins generated significantly higher average profits, indicating that asset selection plays an important role in overall trading performance.


## Key Findings

* Extreme Greed conditions generated the highest trader profitability and win rates.
* Fear markets showed the highest trading activity and largest trade sizes.
* Emotional market conditions created greater volatility in trading outcomes.
* Neutral market conditions produced relatively weaker trading performance.
* Market sentiment significantly influenced trader behavior and risk-taking patterns.


## Conclusion

The analysis demonstrates that Bitcoin market sentiment strongly influences trader behavior, profitability, and risk-taking patterns in the cryptocurrency market.

The findings reveal that traders achieved the highest average profitability and win rates during Extreme Greed conditions, suggesting that bullish market momentum creates favorable trading environments. In contrast, Fear periods recorded the highest trading activity and largest average trade sizes, indicating increased trader participation and aggressive positioning during volatile market conditions.

The visualizations and statistical analysis also showed that emotional market environments lead to greater variability in profit and loss outcomes, while Neutral markets tend to provide fewer profitable opportunities.

Overall, this project highlights the significant relationship between market psychology and trading performance. The results emphasize the importance of sentiment analysis in understanding trader behavior, managing risk, and developing data-driven trading strategies in crypto markets.
